## Create and visualize a problem instance with pre-defined generator

Import benchmark generator:

```python
from bonobench.benchmark_sets import *
```

You can use any generator from `bonobench.benchmark_sets` to generate a problem instance. Here, we generate a problem instance `p` using the `SteppedManySpheres` generator:

```python
p_generator = SteppedManySpheres(2)
p = p_generator.generate_instance(0xC0FFEE)
```

You can evaluate `p` as follows:

```python
x = [-3.1, 4.1]
y = p.evaluate(x)
y
```

We can get a simple dominance visualization like this:

```python
p.dominance_plot()
```

Perform Pareto front and target indicator approximation:

```python
front = p.approximate_indicators()
```

```python
target_hv = front.get_normalized_hv()
target_r2 = front.get_r2()

print(f"Normalized target indicator values: \nHypervolume: {target_hv:.5f} \nR2: {target_r2:.6f}")
front.plot()
```
